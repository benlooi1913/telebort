```python
def find_pairs(numbers, target):
    pairs = []
    num = len(numbers)

    for i in range(numbers):
        for j in range(numbers):
            if numbers[i] + numbers[j] == target:
                pairs.append((numbers[i], numbers[j]))
    return pairs 
```

```python
# The number has been deducted - just need to find complement in seen set
def find_pairs_optimized(numbers, target):
    pairs = []
    seen = set()

    for num in numbers:
        complement = target - num
        if complement in seen:
            pairs.append((complement, num))
        seen.add(num)
    return pairs 

numbers = list(range(100000))
```

```python
# Binary search - O(log n) time complexity, O(1) space complexity

def binary_search(arr, target):
    left, right = 0, len(arr) - 1 
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid 
        elif arr[mid] < target:
            left = mid + 1 
        else:
            right = mid -1 
    return -1 
```

```python
# O(n)
def find_max(arr):
    max_value = arr[0]

    for num in arr:
        if num > max_value:
            max_value = num
    return max_value
```

```python
# O(n log n)
from heapq import merge


def merge_sort(arr):
    if len(arr) <= 1:
        return arr 
    
    mid = len(arr) //2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)
```

```python
# O(n2) - quadratic 

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n - i -1):
            if arr[j] > arr[j +1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    
    return arr 
```
