```python
import cv2
```

```python
# Load trained data 
trained_face_data = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
```

```python
trained_face_data = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye_tree_eyeglasses.xml")
```

```python
image = cv2.imread("outing.jpg")  # Read the image file
resized_image = cv2.resize(image, dsize=(1000,800))  # Resize the image array

if resized_image is None:
    print("Error: Image not found or cannot be loaded.")
else:
    gray = cv2.cvtColor(resized_image, cv2.COLOR_BGR2GRAY) # Perform gray-scaling once
    face_coordinates = trained_face_data.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=10
    )
    # Draw rectangles around faces
    for (x, y, w, h) in face_coordinates:
        cv2.rectangle(resized_image, (x, y), (x+w, y+h), (255, 0, 0), 2)

    for i, (x, y, w, h) in enumerate(face_coordinates, start=1):
        cv2.rectangle(resized_image, (x, y), (x+w, y+h), (255, 0, 0), 2)
        cv2.putText(resized_image, f"Face {i}", (x, y-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

    cv2.imshow("Face Detection", resized_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
    total_detected_faces = len(face_coordinates)
    print(f"Total detected faces: {total_detected_faces}")
```
