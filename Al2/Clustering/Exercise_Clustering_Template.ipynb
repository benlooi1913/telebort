## Question 1 - Simple Dataset

````python
X = np.array([[1, 2], [2, 4], [1.5, 0], [10, 2], [12, 4], [9, 2]])
````

````python
# Import KMeans from sklearn
from sklearn.cluster import KMeans

# Declare the clustering model, kmeans with n_clusters=2 and random_state=42
# Initialize K-Means
kmeans_simple = KMeans(n_clusters=2, random_state=42, n_init='auto')

# Fit the model to the data
kmeans_simple.fit(X)
# Fit the dataset to the model kmeans
````

## Question 1 - Identify a suitable k value

````python
gdown.download('https://drive.google.com/uc?id=148KOIQHyrRson62x1BhWfpWcMMjXk0Zn', 'Live.csv', quiet=False)
live_data = pd.read_csv('Live.csv')
X = live_data[['num_reactions', 'num_comments', 'num_shares','num_likes' , 'num_loves','num_wows' , 'num_hahas', 'num_sads', 'num_angrys']]
````

````python
# Advanced - Question 1: Elbow Method for Optimal K
from sklearn.cluster import KMeans
import pandas as pd
import matplotlib.pyplot as plt

# Declare a list to store the inertia values
list_of_inertia = []

# Use a for loop to loop through K=1 to K=10
for k in range(1, 11):
    # Create a clustering model using the n_clusters = k.
    kmeans = KMeans(n_clusters=k, random_state=42)
    # Fit model with X
    kmeans.fit(X)
    # Append the model's inertia into list_of_inertia
    list_of_inertia.append(kmeans.inertia_)
````

````python
# Advanced - Question 2: Plot Elbow Curve and DataFrame
import pandas as pd
import matplotlib.pyplot as plt

# Create a dataframe to store the cluster number and their respective inertia
elbow_df = pd.DataFrame({'n_clusters': range(1, 11), 'inertia': list_of_inertia})

# Plot the graph of number of clusters against inertia
plt.figure(figsize=(8,5))
plt.plot(elbow_df['n_clusters'], elbow_df['inertia'], marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal K')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
````

Create a model using the ideal k value

There is no fixed answer here as it depends on what k value you choose

````python
# Obtain the k-value from the graph
optimal_k = 3

# Create a model using the selected number of clusters
kmeans_final = KMeans(n_clusters=optimal_k, random_state=42)
kmeans_final.fit(X)
````

````python
# Obtain the inertia of the model
final_inertia = kmeans_final.inertia_
print("Final model inertia:", final_inertia)
````

## Question 2 - Generate Dataset using SKLearn

````python
# Generates a simple dataset
# Dataset contains 400 samples, 3 centers, and a cluster standard deviation of 4.2
from sklearn.datasets import make_blobs
X, _ = make_blobs(n_samples=400, centers=3,
                       cluster_std=4.2, random_state=3)
````

````python
# Plot the dataset into a graph
plt.figure(figsize=(10,7))
plt.scatter(X[:, 0], X[:, 1], s=30);
````

Find the best number of clusters

````python
# Import KMeans from sklearn
from sklearn.cluster import KMeans
import pandas as pd
import matplotlib.pyplot as plt

# Declare a list to store the inertia values
list_of_inertia = []

# Use a for loop to loop through K=1 to K=10
for k in range(1, 11):
    # Create a clustering model using the n_clusters = k.
    kmeans = KMeans(n_clusters=k, random_state=42)
    # Fit model with X
    kmeans.fit(X)
    # Append the model's inertia into list_of_inertia
    list_of_inertia.append(kmeans.inertia_)
````

````python
# Create a dataframe to store the cluster number and their respective inertia
elbow_df = pd.DataFrame({'n_clusters': range(1, 11), 'inertia': list_of_inertia})

# Plot the graph of number of clusters against inertia
plt.figure(figsize=(8,5))
plt.plot(elbow_df['n_clusters'], elbow_df['inertia'], marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal K')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
````

````python
# Obtain the k-value from the graph (for example, let's say k=3 based on the elbow)
k_value = 3

# Create a model using the selected number of clusters
kmeans_final = KMeans(n_clusters=k_value, random_state=42)
kmeans_final.fit(X)
````
