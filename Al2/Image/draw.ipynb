```python
# Download file using requests
import requests

url = "https://drive.google.com/uc?export=download&id=1pPBpeIUp-KEbdiARgLcUy8j7YaTYMMNF"
response = requests.get(url)
with open("stevejob.png", "wb") as f:
    f.write(response.content)
```

```python
import cv2 as cv
import numpy as np
# from google.colab.patches import cv2_imshow

# Read Image
image = cv.imread("stevejob.png")

# Draw Rectangle
# cv.rectangle(image, (start_x, start_y), (end_x, end_y), (B, G, R), thickness)
cv.rectangle(image, (50, 50), (200, 200), (0, 255, 0), 3)

# Draw circle
# cv.circle(image, (center_x, center_y), radius, (B, G, R), thickness)
cv.circle(image, (300, 150), 50, (255, 0, 0), 3)

# Draw line
# cv.line(image, (start_x, start_y), (end_x, end_y), (B, G, R), thickness)
cv.line(image, (0, 0), (400, 300), (0, 0, 255), 5)

# Write Text
# cv.putText(image, 'Text', (x, y), font, fontScale, (B, G, R), thickness, cv.LINE_AA)
cv.putText(image, 'Steve Jobs', (50, 300), cv.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 255), 2, cv.LINE_AA)

# Show image
cv.imshow("Image", image)
cv.waitKey(0)
cv.destroyAllWindows()
```
